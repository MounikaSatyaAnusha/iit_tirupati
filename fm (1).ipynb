```python
import pandas as pd

df = pd.read_csv('feature_matrix_final.csv')

classes = [
    'Utility',
    'Utility_Poly',
    'Water_Body',
    'Water_Body_Pt',
    'RCC_Building',
    'Road',
    'Tin',
    'Tiled'
]

assigned_cols = []

for cls in classes:
    cols = [col for col in df.columns if col.startswith(cls + '_')]
    assigned_cols.extend(cols)
    print(f"{cls}: {len(cols)} columns")

# find missed columns
missed_cols = [col for col in df.columns if col not in assigned_cols and col != 'Tile_Name']

print("\nMissed columns:")
print(missed_cols)
```

```python
import pandas as pd

df = pd.read_csv('feature_matrix_final.csv')

classes = [
    'Utility',
    'Utility_Poly',
    'Water_Body',
    'Water_Body_Pt',
    'RCC_Building',
    'Road',
    'Tin',
    'Tiled'
]

for cls in classes:
    cols = [col for col in df.columns if col.startswith(cls + '_')]
    
    # keep Tile_Name (important)
    class_df = df[['Tile_Name'] + cols]
    
    # save file
    class_df.to_csv(f'{cls}_feature_matrix.csv', index=False)

print("✅ All 8 separate feature matrix CSVs created!")
```
